```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Mall Customer Segmentation Dataset
mall_customers = pd.read_csv('D:\Projects\DATA_ANALYST_PROJECTS\Customer_Segmentation_&_Market_Basket_Analysis\data\Mall_Customers.csv')

# Display first few rows
display(mall_customers.head())
```

```python
print("\nMissing Values:")
print(mall_customers.isnull().sum())
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=mall_customers, palette='Set2')
plt.title("Gender Distribution")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(mall_customers['Age'], bins=20, kde=True, ax=axes[0])
axes[0].set_title("Age Distribution")

sns.histplot(mall_customers['Annual Income (k$)'], bins=20, kde=True, ax=axes[1])
axes[1].set_title("Annual Income Distribution")

sns.histplot(mall_customers['Spending Score (1-100)'], bins=20, kde=True, ax=axes[2])
axes[2].set_title("Spending Score Distribution")

plt.show()
```

```python
sns.pairplot(mall_customers[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']])
plt.show()
```
